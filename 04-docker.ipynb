## 1. The problem: "…but it works on my machine!"

A story you will live through in your career:

> An analyst builds a small web dashboard on her laptop. It works beautifully. She sends it to a colleague — it crashes: *his* laptop has a different Python version. IT installs it on a server — different error: a library is missing. Three days of "works here, breaks there".

Software is fragile about its environment: the app itself, the language version, the libraries, the settings — everything must match. Copying just the app copies *one* of those five things.

**The old shipping industry had the same problem.** Every crate a different shape, loaded by hand, chaos at every harbor. The fix (1956) was almost embarrassingly simple: **the standardized shipping container.** Nobody at the harbor cares what's inside — every container fits every ship, crane, and truck on Earth.

**Docker (2013) is the shipping container for software.** You pack the app *plus everything it needs* into one standardized box. Every computer that can handle boxes (= has Docker) can run it. Identically. Laptop, colleague's laptop, university server, Amazon's cloud — same box, same behavior.

Why *you* should care, as a business student:

- Practically every modern company runs its software in containers — when tech colleagues say "we ship it as a container", you'll know exactly what that means.
- It's the end of installation pain: instead of "install Python, then 12 libraries, then pray", it's *one command* and the thing runs. You will feel this yourself in 30 minutes.

## 2. The four words you need

Docker has its own vocabulary. Master these four and every conversation about containers makes sense — here with a kitchen analogy:

| Word | What it is | Kitchen analogy |
|---|---|---|
| **Dockerfile** | A small text file with build instructions: *"start from X, add Y, run Z"* | The **recipe** 📄 |
| **Image** | The built, frozen package: app + everything it needs. A template — it doesn't "run", it *exists* | The **frozen ready-meal** 🧊 |
| **Container** | A running copy of an image. You can start many containers from one image | The meal, **heated & on the table** 🍽️ |
| **Docker Hub** | A giant online library of ready-made images: databases, web servers, Python, … | The **supermarket** 🏪 |

The flow of everything we do today:

```
 Dockerfile ──[docker build]──▶ Image ──[docker run]──▶ Container
  (recipe)                    (frozen meal)            (meal on the table)
                                  ▲
                                  │  docker pull
                             Docker Hub
                     (supermarket of ready meals)
```

And **Docker Compose**? The dinner-party planner: real systems are *several* containers (app + database + …), and Compose starts and connects them all with one command. Section 8 — the grand finale.

## 4. Your first container 🎉

The traditional first step — run the tiny test image from Docker Hub:

```
docker run hello-world
```

Expected:

```
Unable to find image 'hello-world:latest' locally
latest: Pulling from library/hello-world
c9c5fd25a1bd: Pull complete
Status: Downloaded newer image for hello-world:latest

Hello from Docker!
This message shows that your installation appears to be working correctly.
...
```

Read what happened — it's the whole Docker story in four lines:

1. `Unable to find image … locally` — you asked to run `hello-world`, but there was no such **image** on your machine. **Not an error!**
2. `Pulling from library/hello-world` — so Docker fetched it from **Docker Hub** (the supermarket).
3. Docker created a **container** from the image and ran it.
4. The program inside printed the greeting, finished, and the container stopped.

You have now: pulled an image ✅, run a container ✅. That's `docker run`: *"get the image if needed, then start a container from it."*

## 5. Something real: a web server in one command

`hello-world` is cute; now something genuinely impressive. **Nginx** is one of the web servers behind a big share of the internet. Installing it "properly" takes a sysadmin some effort. With Docker:

```
docker run -d -p 8080:80 --name my-web nginx
```

Expected: a few `Pull complete` lines (first time), then **one long letter-number code** — the container's ID. It's running!

**The proof:** open your normal web browser and go to <http://localhost:8080> — you should see **"Welcome to nginx!"**. A real web server, running in a container on your machine, and your browser is its visitor. (`localhost` = "this computer".)

Decode the command — three flags you'll use constantly:

| Piece | Meaning |
|---|---|
| `-d` | **d**etached — run in the background and give my terminal back (that's why you saw only the ID) |
| `-p 8080:80` | **p**ort mapping — connect *my machine's* door **8080** to the *container's* door **80**. Ports are numbered doors on a computer; nginx listens behind door 80 inside its box, and this flag wires your `localhost:8080` to it. **Left = your side, right = inside the box.** |
| `--name my-web` | a human-friendly name for the container (otherwise Docker invents one like `quirky_einstein` — really) |

### The lifecycle — see, stop, remove

Four commands manage every container's life. Try them in order:

**1. What is running right now?**

```
docker ps
```

Expected — a table with your container:

```
CONTAINER ID   IMAGE   COMMAND    …   PORTS                  NAMES
3f2a9c81b7e4   nginx   "/docker…" …   0.0.0.0:8080->80/tcp   my-web
```

**2. Stop it** (the polite "shut down"):

```
docker stop my-web
```

Check: `docker ps` shows an empty table — and the browser page at `localhost:8080` no longer loads. **Stopped is not gone**, though:

```
docker ps -a
```

The `-a` (all) list still shows `my-web`, status `Exited …` — like a machine switched off but still standing in the room. (You could restart it with `docker start my-web`.)

**3. Remove it** (haul the stopped container away):

```
docker rm my-web
```

Now `docker ps -a` is empty too. The *image* remains on your machine, however — see your local collection:

```
docker images
```

Expected: `nginx` and `hello-world` with their sizes. Images stay until you delete them (`docker rmi nginx` — no need now), so the next `docker run` starts instantly without downloading.

> 💡 **The mental model, once more:** *image* = frozen meal in your freezer (`docker images`), *container* = meal on the table (`docker ps`), `run` = serve a new one, `stop` = done eating, `rm` = clear the table.

> ✅ **Checkpoint** — You ran a real web server with one command, saw it in the browser, and know `docker ps` / `stop` / `rm` / `images`. If `localhost:8080` showed nothing: is the container in `docker ps`? Did you use exactly `-p 8080:80`? See section 10 (troubleshooting).

### The app itself — `app.py`

*(You do **not** need to understand Python — skim it and move on.)*

```python
from flask import Flask, render_template

app = Flask(__name__)


@app.route("/")
def home():
    return render_template("index.html")


if __name__ == "__main__":
    app.run(host="0.0.0.0", port=5000)
```

In words: a mini web server (built with **Flask**, a Python library) that shows the page `templates/index.html` to visitors, listening behind door **5000**.

### The ingredient list — `requirements.txt`

```
Flask==3.1.0
```

One ingredient: the Flask library, in an exact version. (Real apps list dozens here — same idea.)

### The recipe — `Dockerfile`, line by line

This is the star of the show. A `Dockerfile` is a plain text file (named exactly that — no file ending) with one instruction per line:

```dockerfile
# Start from an official image that already contains Python
FROM python:3.12-slim

# Work inside the folder /app in the container
WORKDIR /app

# Copy the ingredient list and install the ingredients
COPY requirements.txt .
RUN pip install --no-cache-dir -r requirements.txt

# Copy the rest of our code into the container
COPY . .

# The app listens on port 5000 inside the container
EXPOSE 5000

# This command starts the app when the container starts
CMD ["python", "app.py"]
```

| Line | Plain English |
|---|---|
| `FROM python:3.12-slim` | *Don't start from zero* — start from a ready-made image with Python 3.12 inside (from the supermarket; "slim" = the compact variant). **Every recipe starts with FROM.** |
| `WORKDIR /app` | Inside the box, create the folder `/app` and work there from now on |
| `COPY requirements.txt .` | Copy the ingredient list from your folder into the box (`.` = "to here", the current folder in the box — Notebook 1!) |
| `RUN pip install …` | Run a command *while building*: install the Python ingredients into the box |
| `COPY . .` | Now copy *everything* in your project folder into the box |
| `EXPOSE 5000` | A label on the box: "the app inside listens behind door 5000" |
| `CMD ["python", "app.py"]` | What the box should do *when a container starts*: launch the app |

> 💡 Why copy `requirements.txt` first and the rest later? Docker is clever about re-building: it re-uses every step that hasn't changed. Ingredients rarely change, your code changes constantly — this order makes rebuilds after code edits nearly instant. A pro habit you get for free.

### Build it

You are standing in `demo-app` (check: `pwd`). Build the image:

```
docker build -t my-first-app .
```

- `-t my-first-app` — **t**ag: the name for your image
- `.` — "the recipe and files are **here**, in the current folder" (that lonely dot is required and forgets-itself easily!)

Expected: Docker walks through the recipe, step by step — you can literally see your Dockerfile executing:

```
[+] Building 12.3s
 => [1/5] FROM docker.io/library/python:3.12-slim
 => [2/5] WORKDIR /app
 => [3/5] COPY requirements.txt .
 => [4/5] RUN pip install --no-cache-dir -r requirements.txt
 => [5/5] COPY . .
 => naming to docker.io/library/my-first-app
```

Check the freezer:

```
docker images
```

**`my-first-app` is in the list.** You just packed your first box. 📦

### Run it

Exactly like nginx — it's a standard box now, and this is the whole point:

```
docker run -d -p 8000:5000 --name my-app my-first-app
```

Read the port mapping like a pro: *your* door **8000** → the container's door **5000** (where the app listens, remember `EXPOSE`). The two numbers **don't have to match** — left is your choice, right is the app's.

**The proof:** browser → <http://localhost:8000> → **"🎉 It works!"**

Two everyday commands while it runs:

```
docker ps
```
→ `my-app` is up, ports `0.0.0.0:8000->5000/tcp`.

```
docker logs my-app
```
→ everything the app printed inside the box — your window into the container. (Flask shows its startup lines and one line per browser visit. This is where you look **whenever a container misbehaves**.)

Then clean up:

```
docker stop my-app
```

```
docker rm my-app
```

## 7. Why one more tool? From `docker run` to Compose

Look at what our little `docker run` line has already grown into:

```
docker run -d -p 8000:5000 --name my-app my-first-app
```

Now imagine a *real* system: the web app **plus** a database **plus** an admin tool — three containers, each with its own flags, started in the right order, wired to each other. Three long commands to remember and retype (in the right order!) every single time, on every machine, by every team member. 😩

**Docker Compose** fixes this: you write the whole setup — every container, port, and setting — **once**, into a file called `compose.yaml`. From then on, the entire system is:

```
docker compose up
```

The file is the orchestra's **sheet music**; Compose is the conductor. As a bonus, the file *is documentation*: anyone can read exactly what the system consists of. (Compose is already installed — it came with Docker in section 3.)

### Our app, as sheet music

You've seen `compose.yaml` sitting in `demo-app` — here is its complete content:

```yaml
services:
  web:
    build: .
    ports:
      - "8000:5000"
```

Line by line:

- `services:` — "here comes the list of containers in my system"
- `web:` — the first (and only) one; the name is your choice
- `build: .` — its image is built from the recipe (Dockerfile) in this folder
- `ports:` — the same port mapping you know, `"your side : box side"`

> ⚠️ **Careful — YAML files are picky about indentation.** Those are exactly **2 spaces** per level (never tabs). When you write your own, copy the structure precisely — a mis-indented line is the #1 compose error.

### Start the system

Standing in `demo-app` (where `compose.yaml` lives — compose commands always work relative to that file):

```
docker compose up
```

Expected: Compose builds if needed, then starts `web` and — since we didn't say `-d` this time — **streams the app's logs live into your terminal**. Check <http://localhost:8000> in the browser: 🎉 It works — and you can watch each visit appear in the log lines. This foreground mode is lovely for watching a system breathe.

Stop it with **Ctrl + C** (the emergency exit from Notebook 1 — here it's the official off-switch).

The background variant, and its controls — the same trio you know from single containers, now for the *whole system*:

```
docker compose up -d
```

```
docker compose ps
```

```
docker compose logs
```

And to stop **and** clean everything away in one move (Compose's `stop` + `rm`):

```
docker compose down
```

## 8. The grand finale — a multi-container system, zero code 🎼

Time for Compose to show off. We'll run a **real database** (PostgreSQL — powers half the startups you know) plus **Adminer**, a web-based database admin tool — two containers, straight from the supermarket, wired together. Amount of code you will write: **none**.

The course repo has the file ready in the `database-demo` folder:

```
cd ~/premaster-digimaster26/database-demo
```

*(No repo? Make an empty folder `database-demo` anywhere and create this `compose.yaml` in it yourself:)*

```yaml
services:
  db:
    image: postgres:16
    environment:
      POSTGRES_PASSWORD: example
    volumes:
      - db_data:/var/lib/postgresql/data

  adminer:
    image: adminer:4
    ports:
      - "8081:8080"

volumes:
  db_data:
```

New things to notice:

- **Two services** — `db` and `adminer`: two containers, one file.
- `image:` instead of `build:` — no recipe of ours; both come ready-made from Docker Hub.
- `environment:` — settings handed into the box at start (here: the database password). This is how containers are configured without touching their insides.
- `volumes:` — where the database keeps its files. Read `db_data:/var/lib/postgresql/data` as: *"attach the storage area called `db_data` to the folder inside the container where PostgreSQL writes."* Containers are built to be thrown away and recreated; anything that has to outlive them goes in a **volume**, which Docker keeps outside the container. The `volumes:` block at the very bottom is where that name is declared — and giving it a **name** is what lets Docker hand back **the same** storage next time. You never create it by hand: `docker compose up` makes it the first time and reuses it ever after. (In Docker's own lists it appears with the project name in front, as `database-demo_db_data`, so volumes from different projects can't collide. You will prove the whole thing to yourself in a moment.)
- Only `adminer` gets a port to *your* machine. The database itself stays unreachable from outside — **but** containers in one compose file share a private network and reach each other **by service name**: to Adminer, the database is simply called `db`. (Yes — container names work like server addresses inside. Notebook 3 vibes.)

### Lift the baton

```
docker compose up -d
```

Expected: `postgres` and `adminer` images download (first time — the database is a few hundred MB, give it a minute), then:

```
[+] Running 4/4
 ✔ Network database-demo_default      Created
 ✔ Volume database-demo_db_data       Created
 ✔ Container database-demo-db-1       Started
 ✔ Container database-demo-adminer-1  Started
```

*(Both there in plain sight: a private `Network` for the two, and the `Volume` that will hold the database's files — created for you.)*

**Now the payoff.** Browser → <http://localhost:8081> — the Adminer login page. Log in **with the values from your compose file**:

| Field | Value | Why |
|---|---|---|
| System | PostgreSQL | that's our `db` service's software |
| Server | `db` | the *service name* — inside the network, that's the database's address! |
| Username | `postgres` | the image's default user |
| Password | `example` | you set it — `POSTGRES_PASSWORD` |
| Database | `postgres` | the default database |

You are now looking at the inside of a live PostgreSQL database — click around, and **do create a table**: hit **Create table**, give it any name and one column, and save. We are about to use it to prove something. **One YAML file, one command** gave you a running database plus its admin console, connected. On any machine. That is why the whole industry runs on this.

When you're done, tear the system down:

```
docker compose down
```

Both containers stop and vanish, network included. **Your data does not** — and that is worth proving to yourself. Start the system again:

```
docker compose up -d
```

Log back in at <http://localhost:8081> and look: **the table you created is still there.** The containers were destroyed and built again from scratch, but the named volume from your compose file was handed straight back to the new one. That is what volumes are for, and it is why every real system that stores anything has one.

And when you *do* want a clean slate, add `-v`:

```
docker compose down -v
```

The `-v` deletes the named volume too — the true factory reset. Start it up after that and the database is empty again.

> 💡 **Small print for later:** containers are meant to be disposable, thrown away and recreated at will — so anything that must outlive them goes in a volume, and that volume gets a **name**. Leave the name out and Docker still makes one, but an anonymous one it cannot match up again: the next `up` would quietly start with an empty database. Named volume = data you keep; no name = clean slate every time.

> ✅ **Checkpoint** — You started a two-container system with one command, logged into the database through the admin tool, watched your data survive a full `down`/`up`, and tore it all down again. You have officially *orchestrated containers*. 🎼

## 9. Housekeeping — Docker and your disk space

Images are big (that's their nature — they carry everything). After months of use, Docker can quietly own many GB of your disk. Your broom cupboard:

| Command | What it does |
|---|---|
| `docker ps -a` | All containers, running or stopped — the first thing to review |
| `docker stop <name>` then `docker rm <name>` | Retire a container |
| `docker images` | All images and their sizes — the space is mostly here |
| `docker rmi <name>` | Delete an image you don't need (re-downloadable anytime) |
| `docker system df` | "How much disk is Docker using, in total?" |
| `docker system prune` | **The big cleanup:** deletes all stopped containers and unused data. It asks `Are you sure?` — read what it lists, then `y` |

> ⚠️ `docker system prune` is safe for everything *running*, but gone is gone for the stopped things it lists. On 🍎/🪟, Docker Desktop's settings also let you cap how much disk Docker may use at all.

One habit is enough: **when you stop working, `docker compose down` / `docker stop` your things — and every few weeks, run a prune.**

## 10. When something goes wrong

| Message / situation | What it means | Fix |
|---|---|---|
| `Cannot connect to the Docker daemon` / `error during connect` | **Docker isn't running.** The #1 issue, by far | Start Docker Desktop, wait for the green whale, retry (🐧: `sudo systemctl start docker`) |
| `docker: command not found` | Install didn't complete, or old terminal | New terminal; recheck section 3. 🪟: also check Docker Desktop → Settings → Resources → **WSL integration** → toggle for Ubuntu **on** → Apply, then open a fresh Ubuntu window |
| `permission denied … docker.sock` (🐧) | Your user isn't in the `docker` group yet | `sudo usermod -aG docker $USER`, then **log out & in** |
| `port is already allocated` / `address already in use` | Something already sits behind that door on *your* machine (often a forgotten container) | `docker ps` → stop the old one; or change the left port number (`-p 8090:80`) — with Compose, change the left number in the `ports:` line of `compose.yaml` instead |
| Browser shows nothing on `localhost:…` | Container not running, or wrong port | `docker ps` (is it up? which ports?), `docker logs <name>` (did it crash?) |
| `pull access denied` / `repository does not exist` | Typo in an image name | Check spelling (`nginx`, `postgres:16`, …) |
| First `docker run`/`pull` takes forever | Images are hundreds of MB — normal once per image | Wait; better Wi-Fi; every later start is instant |
| `WSL 2 installation is incomplete` / virtualization errors (🪟) | WSL missing pieces, or virtualization disabled in the machine's firmware | Admin PowerShell: `wsl --update`, reboot; if it persists, ask the instructors (a firmware/BIOS toggle may be needed) |
| Whale never turns green / Desktop stuck (🍎 🪟) | Docker Desktop had a bad day | Quit fully & reopen; reboot; Docker Desktop → Troubleshoot → Restart |
| `invalid … did not find expected key` on compose | YAML indentation is off | Compare spacing with the examples — 2 spaces, no tabs |
| (🍎) something else already answers on port 5000 | macOS's AirPlay Receiver squats on 5000 | We sidestepped it by publishing on 8000 — do the same in your own projects, or turn AirPlay Receiver off in System Settings |

## 11. Cheat sheet 📋

```
IMAGES (frozen meals)                    CONTAINERS (meals on the table)
  docker build -t name .   build from      docker run -d -p 8000:80 --name x img
                           Dockerfile        run: -d background, -p your:box doors
  docker images            list mine        docker ps          running now
  docker pull name         just download    docker ps -a       incl. stopped
  docker rmi name          delete           docker logs x      what's it saying?
                                            docker stop x      shut down
                                            docker rm x        clear away

COMPOSE (whole systems; needs compose.yaml, run where the file lives)
  docker compose up        start all, watch logs (Ctrl+C stops)
  docker compose up -d     start all, background          docker compose ps
  docker compose down      stop & remove all              docker compose logs

HOUSEKEEPING                             GOLDEN RULES
  docker system df    disk usage           Docker Desktop must be running 🐳
  docker system prune big cleanup (asks)   ports: LEFT = your machine, RIGHT = box
                                           container misbehaving? → docker logs
```

## 1. The problem Git solves

You have seen this folder before — maybe you have *owned* this folder:

```
thesis_final.docx
thesis_final_v2.docx
thesis_final_v2_FINAL.docx
thesis_final_v2_FINAL_profcomments.docx
thesis_REALLY_final_USE_THIS_ONE.docx
```

This "system" has real problems:

- Which version is actually the newest? What changed between them? Nobody knows.
- Two people edit at the same time → someone's work gets overwritten.
- You emailed `v2` to a colleague, then kept editing → now there are two different `v2`s in the world.

Now imagine the same chaos, but for **software**: thousands of files, dozens of people editing simultaneously, for years. Companies needed a real solution.

That solution is **version control**, and **Git** is the version-control tool that won. It gives you:

- 📸 **Named snapshots** of your whole project, whenever *you* decide ("save game" points you can always go back to)
- 🧾 **A full history** — who changed what, when, and *why*
- 🤝 **Safe teamwork** — many people, one project, no overwritten work, no emailing zip files
- ⏪ **An undo button for entire projects** — mistakes stop being scary

Git was created in 2005 by Linus Torvalds (the creator of Linux). Today it is *the* industry standard: if a company builds software, it uses Git. That is why it is worth two hours of your life this week.

## 2. Git vs. GitHub — not the same thing!

This confuses everyone at first, so let's settle it immediately:

| | **Git** | **GitHub** |
|---|---|---|
| What is it? | A **program on your computer** | A **website** (github.com) |
| What does it do? | Tracks versions of a project folder | Stores a shared online copy of Git projects |
| Rough analogy | Track Changes for entire projects | Dropbox + social network for Git projects |
| Do you need an account? | No | Yes (free) |

They work together: **Git** manages the history on your laptop; **GitHub** is the meeting point where a team's copies get synchronized. (GitHub has competitors — GitLab, Bitbucket — same idea, and your Git skills transfer 1-to-1.)

First we master Git *alone, on your machine* (sections 3–7). Then we connect to GitHub (section 8).

## 4. Tell Git who you are (one-time setup)

Every snapshot you take will be stamped with a name and an email — that is how teams see *who* changed *what*. Set yours once (use your real name and the email you will use for GitHub; **type your own details, not "Anna's"**):

```
git config --global user.name "Anna Example"
```

```
git config --global user.email "anna.example@student.university.edu"
```

And one setting that makes Git's behavior modern and consistent everywhere:

```
git config --global init.defaultBranch main
```

Check your work:

```
git config --global --list
```

Expected — your details echoed back:

```
user.name=Anna Example
user.email=anna.example@student.university.edu
init.defaultbranch=main
```

(If the list is long and the prompt seems gone, you are in a *viewer* — press `q` to quit. Notebook 1, survival skills!)

> 💡 **Tip** — `--global` means "for everything I do on this computer", so you do this exactly once, not per project.

## 5. How Git thinks — the photo analogy 📸

Git will make much more sense with one mental picture. In every Git project, your files exist in **three zones**:

```
 ┌───────────────┐   git add    ┌───────────────┐   git commit   ┌────────────────┐
 │ WORKING FOLDER │ ──────────▶ │ STAGING AREA  │ ─────────────▶ │    HISTORY     │
 │  the files you │             │ "who is in    │                │ the photo album│
 │  see and edit  │             │  the photo?"  │                │  of snapshots  │
 └───────────────┘             └───────────────┘                └────────────────┘
```

Think of taking a **group photo**:

1. **Working folder** — the room where everyone walks around. These are the normal files you see and edit. Git watches, but nothing is recorded yet.
2. **Staging area** — you arrange *who will be in the photo*. `git add file` means: *"this file's current state should be part of the next snapshot."* You can add several files — composing the photo.
3. **History** — `git commit` **takes the photo**. The snapshot (called a **commit**) goes into the album, permanently, with your name, the date, and a message describing it.

Why the extra staging step? Because it lets you snapshot *related changes together* ("photo of the marketing team only"), even if you edited ten other files too. For now, just remember the rhythm you will use hundreds of times:

**edit → `git add` → `git commit`** *(change things → choose what's in the photo → take the photo)*

## 6. Hands-on — your first repository

A project folder managed by Git is called a **repository** ("repo" for short). Let's make one. Open your terminal:

```
cd ~
```

```
mkdir digimaster_git
```

```
cd digimaster_git
```

Now the magic word that turns a plain folder into a repository:

```
git init
```

Expected:

```
Initialized empty Git repository in /Users/anna/digimaster_git/.git/
```

What just happened? Git created a **hidden subfolder called `.git`** — see it with `ls -a` (Windows: `ls -Force`). That hidden folder *is* the photo album: all snapshots and history will live in there. Two rules:

- **Never edit or delete `.git`** — deleting it deletes the project's entire history (the normal files stay).
- Everything inside `digimaster_git` is now watched by Git. Folders *outside* are not — Git only manages folders where you ran `git init`.

### Step 1 — meet your most-used command: `git status`

```
git status
```

Expected:

```
On branch main

No commits yet

nothing to commit (create/copy files and use "git add" to track)
```

`git status` is Git answering *"what is the current situation?"*. You will type it constantly — before and after almost everything. It always tells you what Git sees and usually even suggests what to do next. **When in doubt: `git status`.**

### Step 2 — create a file and see Git notice it

Make a simple file (Notebook 1 skills — same command on all systems):

```
echo "milk" > groceries.txt
```

```
git status
```

Expected (shortened):

```
Untracked files:
  (use "git add <file>..." to include in what will be committed)
        groceries.txt
```

**Untracked** = *"there is a new file in the room, but it has never been in any photo."* Git sees it but does not record it until you say so.

### Step 3 — stage it (arrange the photo)

```
git add groceries.txt
```

Silence — which, as you know, means success. Ask for the situation:

```
git status
```

Expected:

```
Changes to be committed:
        new file:   groceries.txt
```

The file is now **staged**: standing in frame, ready for the snapshot.

### Step 4 — commit (take the photo!)

```
git commit -m "Add grocery list"
```

`-m` attaches the **message** — a short human sentence saying what this snapshot is. Expected:

```
[main (root-commit) 1a2b3c4] Add grocery list
 1 file changed, 1 insertion(+)
 create mode 100644 groceries.txt
```

🎉 **That was your first commit** — a permanent, named snapshot of the project.

> 💡 **Good commit messages** describe the change, short and in the present tense: `Add grocery list`, `Fix typo in intro`, `Update sales figures for Q3`. Six months from now, messages like `stuff` or `asdf` will help nobody — including you.
>
> ⚠️ **Careful** — if you forget `-m "..."`, Git drops you into a strange full-screen editor asking for a message. If that happens and it looks alien (`~` symbols everywhere = the editor Vim): press **Esc**, type `:q!`, press **Enter** — you're out, nothing happened, redo the commit *with* `-m`.

### Step 5 — look at the album

```
git log
```

Expected:

```
commit 1a2b3c4d5e6f7...(a long letter-number code)
Author: Anna Example <anna.example@student.university.edu>
Date:   Tue Sep 8 10:15:23 2026 +0200

    Add grocery list
```

Reading it:

- **commit 1a2b3c...** — the snapshot's unique fingerprint (Git generates it; people use the first 7 characters to refer to it)
- **Author** — the name/email from section 4 (this is why we configured it!)
- **Date** and your **message**

If the log is long, Git shows it in a viewer — scroll with ↑/↓ and **quit with `q`**. For a compact view, try `git log --oneline`.

### Step 6 — change something and see what Git makes of it

```
echo "eggs" >> groceries.txt
```

(Remember: `>>` *appends* a line.) Now:

```
git status
```

Expected: `modified: groceries.txt` — Git instantly knows the file differs from the last photo. But *what exactly* changed? That is:

```
git diff
```

Expected (core of it):

```
 milk
+eggs
```

Lines starting with `+` are new (shown green in most terminals); deleted lines would start with `-`. `git diff` = *"show me exactly what I changed since the last snapshot"* — press `q` if it opens in the viewer.

### Step 7 — snapshot the change

The rhythm you already know — add, commit, and look at the album:

```
git add groceries.txt
```

```
git commit -m "Add eggs to grocery list"
```

```
git log --oneline
```

Expected:

```
9z8y7x6 (HEAD -> main) Add eggs to grocery list
1a2b3c4 Add grocery list
```

Two photos in the album, newest on top. (`HEAD -> main` is Git-speak for "you are here".)

### Step 8 — the payoff: Git as your undo button ⏪

Let's simulate a disaster. "Accidentally" wreck your file — note the single `>`, which **overwrites**:

```
echo "OOPS everything is gone" > groceries.txt
```

```
cat groceries.txt
```

Expected — your list is destroyed:

```
OOPS everything is gone
```

Before Git, this was a real loss. Now? One command — *"restore this file from the last snapshot"*:

```
git restore groceries.txt
```

```
cat groceries.txt
```

Expected:

```
milk
eggs
```

**Everything is back.** This is why professionals commit early and often: with a good photo album, no mistake in the working folder is ever fatal.

> 💡 There are also commands to bring back *any older* version of any file, or the whole project as it was last Tuesday — beyond our scope today, but that safety net is what you are building every time you commit.

> ✅ **Checkpoint** — You have: turned a folder into a repository (`git init`), checked the situation (`git status`), staged (`git add`), committed twice (`git commit -m`), read the history (`git log`), inspected a change (`git diff`), and undone a mistake (`git restore`). That is the daily core of Git — everything else is built on this loop.

### What GitHub adds to Git

Your repository so far lives only on your laptop. GitHub hosts a **shared online copy** (called a **remote**) that a whole team can synchronize with:

```
        GitHub (the shared copy, in the cloud)
             ▲                    │
        git push              git clone / git pull
       (send my new           (get the repo /
        snapshots up)          get new snapshots down)
             │                    ▼
   your laptop        ◀──▶   your teammates' laptops
```

Three new verbs, and you know the whole picture:

| Command | Meaning |
|---|---|
| `git clone <address>` | Download a full copy of a repository — **including its entire photo album** — onto your computer (done once) |
| `git pull` | *"Fetch whatever is new"* from GitHub into your copy (done all the time) |
| `git push` | *"Upload my new snapshots"* to GitHub so others can pull them (needs permission) |

### A note on `git push` (and why we wait one notebook)

`git push` uploads *your* snapshots to a repository on GitHub — for your own repos or projects where you are invited as a collaborator.

But GitHub needs to be sure it is really *you* — and it does **not** accept your account password from the command line. It wants a cryptographic proof of identity: an **SSH key**. You don't have one yet.

**That is exactly what Notebook 3 is about.** At the end of it, you will connect your key to GitHub and be able to push. (Everything you did today — clone and pull from public repositories — works without any of that.)

## 9. Optional bonus — branches in five minutes 🌿

*Skip this on first reading if you are tired; it is here because you will hear the word constantly.*

A **branch** is a **parallel version of the project** — like duplicating your photo album to experiment freely, while the original (`main`) stays untouched. Teams do all real work on branches and merge them into `main` when ready.

Try the miniature version in your `digimaster_git` repo (`cd ~/digimaster_git` first):

| Command | What happens |
|---|---|
| `git branch` | List branches — you have `main`, the `*` marks where you are |
| `git switch -c experiment` | **C**reate branch `experiment` and move onto it |
| `echo "cheese" >> groceries.txt` then `git add groceries.txt` and `git commit -m "Try adding cheese"` | A commit that exists **only on `experiment`** |
| `cat groceries.txt` | cheese is there |
| `git switch main` | Back to the main branch… |
| `cat groceries.txt` | …and cheese is *gone* — `main` never saw that commit. Parallel worlds! |
| `git merge experiment` | Pull the experiment's snapshots into `main` — cheese appears (`cat groceries.txt`) |

That is the entire mental model: *branch = safe parallel copy, merge = bring it home.* Where two people change the *same line* in parallel, merging asks a human to decide — that is the famous **merge conflict**: not an error, just Git saying "you decide". How to resolve those is a story for your future teams.

## 11. When something goes wrong

| Message / situation | What it means | Fix |
|---|---|---|
| `fatal: not a git repository` | You are in a folder without a `.git` — Git commands only work *inside* repos | `pwd`, then `cd` into your project folder |
| `Please tell me who you are` | Name/email not configured | Do section 4 |
| `nothing to commit, working tree clean` | No changes since the last snapshot — nothing to photograph | Not an error! Change a file first, or check `git status` |
| `Changes not staged for commit` after committing | You edited but forgot `git add` before `git commit` | `git add file`, then commit again |
| A full screen of `~` symbols appeared | You forgot `-m` and landed in the Vim editor | **Esc**, then `:q!`, then **Enter** — and commit again with `-m "..."` |
| `git log` / `git diff` trapped me | It's the pager/viewer | Press `q` |
| `Repository not found` when cloning | Typo in the address, or the repo is private | Check the URL character by character |
| `CONFLICT` during merge/pull | Two versions changed the same lines; Git wants a human decision | Don't panic, nothing is lost — for this course, ask an instructor |

> 💡 Also genuinely professional: googling error messages. Every developer does it daily. Copy the message, add the word "git", and the first results usually solve it.
